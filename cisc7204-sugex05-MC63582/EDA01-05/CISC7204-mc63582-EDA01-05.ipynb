## Q.1-1) Find all the unique 'Wind Speed' values in the data.


```python
print('Number of unique Wind Speed values:', data['Wind Speed_km/h'].nunique())
print(data['Wind Speed_km/h'].unique())
```

## Q.1-2) Find the number of times when the 'Weather is exactly Clear'.


```python
# Method 1: value_counts
print(data['Weather'].value_counts())

# Method 2: filtering
print('\nExactly Clear records:', data[data.Weather == 'Clear'].shape[0])
data[data.Weather == 'Clear']
```

## Q.1-3) Find the number of times when the 'Wind Speed was exactly 4 km/h'.


```python
print('Wind Speed exactly 4 km/h:', (data['Wind Speed_km/h'] == 4).sum())
data[data['Wind Speed_km/h'] == 4]
```

## Q.1-4) Find out all the Null Values in the data.


```python
print(data.isnull().sum())
print('\nTotal null cells:', data.isnull().sum().sum())
data.notnull().sum()
```

## Q.1-5) Rename the column name 'Weather' of the dataframe to 'Weather Condition'.


```python
data.rename(columns={'Weather': 'Weather Condition'}, inplace=True)
data.head()
```

## Q.1-6) What is the mean 'Visibility' ?


```python
print('Mean Visibility (km):', data['Visibility_km'].mean())
```

## Q.1-7) What is the Standard Deviation of 'Pressure' in this data?


```python
print('Std of Press_kPa:', data['Press_kPa'].std())
```

## Q.1-8) What is the Variance of 'Relative Humidity' in this data ?


```python
print('Variance of Rel Hum_%:', data['Rel Hum_%'].var())
```

## Q.1-9) Find all instances when 'Snow' was recorded.


```python
# Exactly 'Snow'
print('Weather Condition == Snow:', (data['Weather Condition'] == 'Snow').sum())

# Any record that mentions Snow (Snow Showers, Freezing Rain,Snow, ...)
snow_records = data[data['Weather Condition'].str.contains('Snow', na=False)]
print('Records containing Snow:', snow_records.shape[0])
snow_records
```

## Q.1-10) Find all instances when 'Wind Speed is above 24' and 'Visibility is 25'.


```python
cond = (data['Wind Speed_km/h'] > 24) & (data['Visibility_km'] == 25)
print('Matching records:', cond.sum())
data[cond]
```

## Q.1-11) What is the Mean value of each column against each 'Weather Condition' ?


```python
data.groupby('Weather Condition').mean(numeric_only=True)
```

## Q.1-12) What is the Minimum & Maximum value of each column against each 'Weather Condition' ?


```python
print('----- Minimum -----')
display(data.groupby('Weather Condition').min(numeric_only=True))
print('----- Maximum -----')
data.groupby('Weather Condition').max(numeric_only=True)
```

## Q.1-13) Show all the Records where Weather Condition is Fog.


```python
fog = data[data['Weather Condition'] == 'Fog']
print('Fog records:', fog.shape[0])
fog
```

## Q.1-14) Find all instances when 'Weather is Clear' or 'Visibility is above 40'.


```python
cond = (data['Weather Condition'] == 'Clear') | (data['Visibility_km'] > 40)
print('Matching records:', cond.sum())
data[cond]
```

## Q.1-15) Find all instances when:
A. 'Weather is Clear' and 'Relative Humidity is greater than 50'  
**or**  
B. 'Visibility is above 40'


```python
cond = ((data['Weather Condition'] == 'Clear') & (data['Rel Hum_%'] > 50)) | (data['Visibility_km'] > 40)
print('Matching records:', cond.sum())
data[cond]
```
