## 0. Load the data and inspect the dataframe

## Q.2-1) Instruction (For Data Cleaning)
Find all Null Values in the dataset. If there is any null value in any column, then fill it with the mean of that column.


```python
print('Null values before filling:')
print(car.isnull().sum())
```

```python
# Fill every numeric column's nulls with that column's mean.
# Object columns (Make, Model, ...) have no numeric mean, so they stay as-is.
numeric_cols = car.select_dtypes(include='number').columns
car[numeric_cols] = car[numeric_cols].fillna(car[numeric_cols].mean())

print('Null values after filling numeric columns with mean:')
print(car.isnull().sum())
```

## Q.2-2) Question (Based on Value Counts)
Check what are the different types of Make are there in our dataset. And, what is the count (occurrence) of each Make in the data?


```python
print('Number of different Makes:', car['Make'].nunique())
car['Make'].value_counts()
```

## Q.2-3) Instruction (Filtering)
Show all the records where Origin is Asia or Europe.


```python
asia_europe = car[car['Origin'].isin(['Asia', 'Europe'])]
print('Asia or Europe records:', asia_europe.shape[0])
asia_europe
```

## Q.2-4) Instruction (Removing unwanted records)
Remove all the records (rows) where Weight is above 4000.


```python
print('Rows before removal:', car.shape[0])
print('Rows with Weight > 4000:', (car['Weight'] > 4000).sum())

car = car[~(car['Weight'] > 4000)]
print('Rows after removal:', car.shape[0])
car
```

## Q.2-5) Instruction (Applying function on a column)
Increase all the values of 'MPG_City' column by 3.


```python
print('MPG_City before:')
print(car['MPG_City'].head())

car['MPG_City'] = car['MPG_City'].apply(lambda x: x + 3)

print('\nMPG_City after +3:')
print(car['MPG_City'].head())
car.head()
```
