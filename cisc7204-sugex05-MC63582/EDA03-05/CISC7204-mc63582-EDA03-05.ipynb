## Q.3-1) Instruction (For Data Cleaning)
Remove the column that only contains missing values.


```python
# country_name is 100% missing
print(police['country_name'].isnull().all())
police.drop(columns='country_name', inplace=True)
police.head()
```

## Q.3-2) Question (Based on Filtering + Value Counts)
For Speeding, were Men or Women stopped more often?


```python
speeding = police[police['violation'] == 'Speeding']
print(speeding['driver_gender'].value_counts())
print('\nMen were stopped more often for Speeding.')
```

## Q.3-3) Question (Groupby)
Does gender affect who gets searched during a stop?


```python
print('Search rate by gender:')
print(police.groupby('driver_gender')['search_conducted'].sum())
print()
print('Overall searches:', police['search_conducted'].sum())
print()
print('Mean search rate by gender:')
police.groupby('driver_gender')['search_conducted'].mean()
```

## Q.3-4) Question (mapping + data-type casting)
What is the mean stop_duration?


```python
print(police['stop_duration'].value_counts())

# Map duration bins to representative minutes, then take the mean
police['stop_duration'] = police['stop_duration'].map(
    {'0-15 Min': 7.5, '16-30 Min': 23, '30+ Min': 45}
)
print('\nMapped stop_duration value counts:')
print(police['stop_duration'].value_counts())
print('\nMean stop_duration (minutes):', police['stop_duration'].mean())
```
