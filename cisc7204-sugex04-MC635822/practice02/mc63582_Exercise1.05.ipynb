```python
# 01 导入 pandas 库
import pandas as pd

# 02 加载数据集，以第一列（国家名）作为索引
dataset = pd.read_csv('world_population.csv', index_col=0)
```

```python
# 03 打印前两行
dataset[0:2]
```

```python
# 04 计算第三行（索引为2）的平均值，axis=1 按行
dataset.iloc[[2]].mean(axis=1, numeric_only=True)
```

```python
# 05 计算最后一行的平均值
dataset.iloc[[-1]].mean(axis=1, numeric_only=True)
```

```python
# 06 计算德国所在行的平均值
dataset.loc[["Germany"]].mean(axis=1, numeric_only=True)
```

```python
# 07 计算最后一行的中位数
dataset.iloc[[-1]].median(axis=1, numeric_only=True)
```

```python
# 08 计算最后三列（columns）的中位数
# 用 iloc[:, -3:] 取最后三列，再对每行计算中位数
dataset.iloc[:, -3:].median(axis=1, numeric_only=True)
```

```python
# 09 计算前10行的中位数
dataset.head(10).median(axis=1, numeric_only=True)
```

```python
# 10 计算方差，返回最后五列
dataset.var(numeric_only=True).tail()
```

```python
# 11 计算2015年的平均值，分别用 pandas 和 numpy
import numpy as np
print("pandas", dataset["2015"].mean())
print("numpy", np.mean(dataset["2015"]))
```
