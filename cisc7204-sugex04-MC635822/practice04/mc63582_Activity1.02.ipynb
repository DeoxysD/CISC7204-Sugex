```python
# 01 导入 pandas
import pandas as pd

# 02 加载森林火灾数据集
forest = pd.read_csv('forestfires.csv')
```

```python
# 03 打印前两行，查看数据结构
forest.head(2)
```

```python
# 04 筛选 area > 0 的数据
forest_fires = forest[forest["area"] > 0]
```

```python
# 05 计算 area 列的统计值
forest_fires["area"].mean()
forest_fires["area"].min()
forest_fires["area"].max()
forest_fires["area"].std()
```

```python
# 06 按 area 排序，打印最后20行（最大的火灾）
forest_fires.sort_values(by=["area"]).tail(20)
```

```python
# 07 计算 area 的中位数，与均值对比
forest_fires["area"].median()
```

```python
# 08 获取 month 列的唯一值列表
forest["month"].unique()
```

```python
# 09 统计三月的火灾数量
forest[forest["month"] == "mar"].shape
```

```python
# 10 遍历所有月份，统计火灾数、平均温度
for month in forest["month"].unique():
    month_data = forest[forest["month"] == month]
    count = month_data.shape[0]
    avg_temp = month_data["temp"].mean()
    print(f"Month: {month}, Fires: {count}, Avg Temp: {avg_temp:.2f}")
```
